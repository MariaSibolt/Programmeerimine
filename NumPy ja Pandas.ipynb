```python
import numpy as np

A = np.array([[4, 5, -2], [3, -1, 0], [4, 2, 7]])
B = np.array([[2, 1, -1], [0, 1, 3], [5, 7, 3]])

result = 3 * A.T - (A + 2 * B).T
print(result)
```

```python
A = np.array([[-2, 3, 2], [1, 4, 1]])
B = np.array([[1, 2], [0, 5], [3, -1]])

result = 2 * A.T - 3 * B
print(result)
```

```python
A = np.array([[-1, 1], [0, 2]])
B = np.array([[-1, 1, 3, 0], [0, 2, -2, -1]])

BBT = B @ B.T
result = 2 * (A - BBT) + A.T
print(result)
```

```python
A = np.array([[1, -2], [-3, 4]])
B = np.array([[1, 2], [5, 1], [3, -1]])

A_squared = A @ A
BTB = B.T @ B
result = (A_squared - BTB) @ A
print(result)
```

SymPy

```python
import sympy

x, y, z = sympy.symbols('x y z')

eq1 = sympy.Eq(6*x - 3*y, 4)
eq2 = sympy.Eq(6*x + 5*y, 4)
sol1 = sympy.solve((eq1, eq2), (x, y))
print("2x2 süsteem 1:", sol1)
```

```python
eq3 = sympy.Eq(-x + 7*y, 5)
eq4 = sympy.Eq(-3*x + 2*y, 0)
sol2 = sympy.solve((eq3, eq4), (x, y))
print("2x2 süsteem 2:", sol2)
```

```python
eq5 = sympy.Eq(3*x + 2*y + z, 5)
eq6 = sympy.Eq(2*x + 3*y + z, 1)
eq7 = sympy.Eq(2*x + y + 3*z, 11)
sol3 = sympy.solve((eq5, eq6, eq7), (x, y, z))
print("3x3 süsteem 1:", sol3)
```

```python
eq8 = sympy.Eq(x + 2*y + 4*z, 31)
eq9 = sympy.Eq(5*x + y + 2*z, 29)
eq10 = sympy.Eq(3*x - y + z, 10)
sol4 = sympy.solve((eq8, eq9, eq10), (x, y, z))
print("3x3 süsteem 2:", sol4)
```

1. Nutika hoone inimeste liikumise analüüs (RFID-andmed)

```python
import pandas as pd
import random
from datetime import datetime, timedelta
employee = {1056: 'IT', 2043: 'Müük', 3178: 'Logistika', 2078: 'HR', 4025: 'Turvalisus'}

start_date = datetime(2025, 4, 9, 8, 0)
entries = []

n = 100
while len(entries) < n:
    for emp in employee:
        department = employee[emp]
        begin_date = start_date + timedelta(minutes=random.randint(0, 60))
        end_date = begin_date + timedelta(minutes=random.randint(240, 480))
        entries.append([begin_date.strftime('%Y-%m-%d %H:%M'), emp, 'Sisenemine', department])
        entries.append([end_date.strftime('%Y-%m-%d %H:%M'), emp, 'Väljumine', department])
    start_date = start_date - timedelta(days=1)

df = pd.DataFrame(entries, columns=['Timestamp', 'Employee_ID', 'Access_Point', 'Department'])
df.head(10)
```

```python
n = 100
start_date = datetime(2025, 4, 9, 8, 0)
entries = []
target_anomalies = 10
anomaly_count = 0

while len(entries) < n:
    for emp in employee:
        department = employee[emp]
        begin_date = start_date + timedelta(minutes=random.randint(0, 60))
        end_date = begin_date + timedelta(minutes=random.randint(240, 480))
        
        if anomaly_count < target_anomalies and random.random() < 0.4:
            anomaly_type = random.choice(['no_exit', 'night_entry', 'double_entry'])
            
            if anomaly_type == 'no_exit':
                entries.append([begin_date.strftime('%Y-%m-%d %H:%M'), emp, 'Sisenemine', department])
                anomaly_count += 1
            
            elif anomaly_type == 'night_entry':
                night_time = start_date - timedelta(minutes=random.randint(360, 480))
                entries.append([night_time.strftime('%Y-%m-%d %H:%M'), emp, 'Sisenemine', department])
                entries.append([night_time.strftime('%Y-%m-%d %H:%M'), emp, 'Väljumine', department])
                anomaly_count += 1
            
            elif anomaly_type == 'double_entry':
                entries.append([begin_date.strftime('%Y-%m-%d %H:%M'), emp, 'Sisenemine', department])
                second_entry = begin_date + timedelta(minutes=random.randint(120, 240))
                entries.append([second_entry.strftime('%Y-%m-%d %H:%M'), emp, 'Sisenemine', department])
                anomaly_count += 1
        else:
            entries.append([begin_date.strftime('%Y-%m-%d %H:%M'), emp, 'Sisenemine', department])
            entries.append([end_date.strftime('%Y-%m-%d %H:%M'), emp, 'Väljumine', department])
        
    start_date = start_date - timedelta(days=1)

df = pd.DataFrame(entries, columns=['Timestamp', 'Employee_ID', 'Access_Point', 'Department'])
df.head(10)
```

```python
df.to_csv('access_log.csv', index=False, encoding='utf-8')
print('Fail "access_log.csv" on loodud')
```

```python
df['Timestamp'] = pd.to_datetime(df['Timestamp'])

df['Date'] = df['Timestamp'].dt.date
df['Hour'] = df['Timestamp'].dt.hour
df['Minute'] = df['Timestamp'].dt.minute

df.head()
```

Hoones viibivate inimeste arv igal tunnil 

```python
df['Change'] = np.where(df['Access_Point'] == 'Sisenemine', 1, -1)

entries = df[df['Change'] == 1].groupby(['Date', 'Hour']).size()

exits = df[df['Change'] == -1].groupby(['Date', 'Hour']).size()

people_by_hour = df.groupby(['Date', 'Hour'])['Change'].sum().cumsum()

summary = pd.DataFrame({
    'Sisenemised': entries,
    'Väljumised': exits,
    'Netoliikumine' : people_by_hour
}).fillna(0).astype(int)
summary
```

Töötajad, kes viibisid hoones kõige kauem

```python
durations = {}

for i in range(len(df) - 1):
    row_in = df.iloc[i]
    row_out = df.iloc[i + 1]
    if row_in['Employee_ID'] == row_out['Employee_ID'] \
        and row_in['Access_Point'] == 'Sisenemine' \
        and row_out['Access_Point'] == 'Väljumine':
        duration = round((row_out['Timestamp'] - row_in['Timestamp']).total_seconds() / 3600, 1)
        emp = row_in['Employee_ID']
        durations[emp] = durations.get(emp, 0) + duration
        
result = pd.Series(durations).sort_values(ascending=False)

print('Töötajad, kes viibisid hoones kõige kauem:')
print(result)
```

Osakond, mille töötajad viibisid hoones kõige kauem

```python
department_total = {}

for emp_id, duration in result.items():
    department = employee.get(emp_id, 'Tundmatu')
    department_total[department] = department_total.get(department, 0) + duration
    
department_result = pd.Series(department_total).sort_values(ascending=False)
print('Osakonnad, mille töötajad viibisid hoones kõige kauem:')
print(department_result)
```

Tööpäeva pikkust – kes töötas üle 10 tunni?

```python
workdays = df.groupby(['Employee_ID', 'Date'])['Timestamp'].agg(['first', 'last'])

workdays['Duration'] = round((workdays['last'] - workdays['first']).dt.total_seconds() / 3600, 2)

over_10h = workdays[workdays['Duration'] > 10]
over_10h['Department'] = over_10h.index.get_level_values('Employee_ID').map(employee)

if over_10h.empty:
    print('Ühtegi tööpäeva üle 10 tunni ei leitud.')
else:
    print('üle 10h tööpäevad:')
    print(over_10h)
```

Tipptunnid ja millal hoone on peaaegu tühi

```python
summary = pd.DataFrame({
    'Sisenemised': entries,
    'Väljumised': exits,
    'Netoliikumine': people_by_hour
}).fillna(0).astype(int)

# tipphetk ja tühjeim hetk
busiest = summary['Netoliikumine'].idxmax()
quietest = summary['Netoliikumine'].idxmin()

# kuupäev + tund
busy = datetime.combine(busiest[0], datetime.min.time()) + timedelta(hours=int(busiest[1]))
quiet = datetime.combine(quietest[0], datetime.min.time()) + timedelta(hours=int(quietest[1]))
print("Tipphetk:")
print(f"{busy.strftime('%Y-%m-%d %H:%M')} - {summary.loc[busiest]['Netoliikumine']} inimest")
print("\nTühjeim hetk:")
print(f"{quiet.strftime('%Y-%m-%d %H:%M')} - {summary.loc[quietest]['Netoliikumine']} inimest")
```

```python
import pandas as pd

df = pd.read_csv("smart_home.csv")

#'time' datetime formaati ja indeksiks
df['time'] = pd.to_datetime(df['time'])
df = df.set_index('time')

house_overall_col = "House overall [kW]"
house_average_overall = df[house_overall_col].mean()
min_house_overall = df[house_overall_col].min()
max_house_overall = df[house_overall_col].max()
aeg = df[house_overall_col].idxmax()


print("Üldine energiatarbimine:")
print(f"Keskmine: {house_average_overall:.4f} kW")
print(f"Minimaalne: {min_house_overall:.5f} kW")
print(f"Maksimaalne: {max_house_overall:.3f} kW")
print(f"Aeg, mil energiatarbimine oli maksimaalne: {aeg}\n")
```

1. Keemiliste elementide perioodilisuse tabel 

```python
import csv
import re
class ChemicalElement:
    """
    Keemilise elemendi klass. Kõiki CSV-failist loetud andmeid sõnastikus.

    Args:
        data (dict): Sõnastik, mis sisaldab ühe elemendi kõiki atribuute.
    """
    def __init__(self, data: dict):
        self.data = data

    def __str__(self):
        """
        Tagastab stringina kõik elemendi andmed eraldi ridadel.

        Returns:
            str: Vormindatud tekst elemendi kõigi omadustega.
        """
        return "\n".join(f"{key}: {value}" for key, value in self.data.items())

    def get(self, key):
        """
        Võimaldab sõnastikust väärtust saada kindla võtme alusel.

        Args:
            key (str): Atribuudi nimi, mille väärtust soovitakse.

        Returns:
            Any: Vastav väärtus või None, kui ei leitud.
        """
        return self.data.get(key)


class ChemicalElementService:
    """
    Haldab keemiliste elementide laadimist, otsimist ja arvutusi.

    Args:
        filename (str): CSV-faili nimi, kust andmed laaditakse.
    """
    def __init__(self, filename):
        self.elements = []
        self.symbol_map = {}
        self.load_elements(filename)

    def load_elements(self, filename):
        """
        Laeb kõik elemendid CSV-failist ja salvestab need listi ja sümbolikaardile.

        Args:
            filename (str): Failitee.
        """
        with open(filename, newline='', encoding='utf-8') as csvfile:
            reader = csv.DictReader(csvfile)
            for row in reader:
                element = ChemicalElement(row)
                self.elements.append(element)
                symbol = row.get('Symbol')
                if symbol:
                    self.symbol_map[symbol] = element

    def find_by_name(self, name):
        """
        Otsib elementi nime järgi (nt 'Oxygen').

        Args:
            name (str): Elemendi nimi.

        Returns:
            ChemicalElement or None: Vastav element või None kui ei leitud.
        """
        for element in self.elements:
            if element.get('Element') and element.get('Element').lower() == name.lower():
                return element
        return None

    def calculate_molecular_mass(self, formula):
        """
        Arvutab keemilise valemi molekulaarmassi

        Args:
            formula (str): Keemiline valem

        Returns:
            float: Molekulaarmass ühikutes

        Raises:
            ValueError: Kui valemis on tundmatu element
        """
        pattern = r'([A-Z][a-z]*)(\d*)'
        matches = re.findall(pattern, formula)
        total_mass = 0.0
        for symbol, count in matches:
            if symbol not in self.symbol_map:
                raise ValueError(f"Tundmatu element valemis: {symbol}")
            count = int(count) if count else 1
            weight = self.symbol_map[symbol].get('Weight')
            if weight:
                total_mass += float(weight) * count
        return total_mass


def main():
    """
    Kuvab menüü ja ootab kasutaja sisendeid
    """
    service = ChemicalElementService("elements (1).csv")

    while True:
        print("\nKeemiliste elementide perioodilisustabel")
        print("1. Otsi elementi nime järgi ja kuva kõik andmed")
        print("2. Arvuta molekulaarmass (nt H2O)")
        print("3. Välju")
        choice = input("Valik (1-3): ")

        if choice == '1':
            name = input("Sisesta elemendi nimi inglise keeles (nt Hydrogen): ")
            element = service.find_by_name(name)
            if element:
                print("\n" + str(element))
            else:
                print("Elementi ei leitud.")
        elif choice == '2':
            formula = input("Sisesta molekulaarvalem (nt H2O): ")
            try:
                mass = service.calculate_molecular_mass(formula)
                print(f"\nMolekulaarmass: {mass:.3f} u")
            except ValueError as e:
                print("Viga:", e)
        elif choice == '3':
            print("Programm lõpetati.")
            break
        else:
            print("Vigane valik. Proovi uuesti.")


if __name__ == "__main__":
    main()
```

2. Tarkvara

```python
from datetime import datetime
from typing import List

class Software:
    """
    Tarkvara baasklass.

    Args:
        name (str): Tarkvara nimi.
        producer (str): Tarkvara tootja.
        install_date (str või datetime.date): Tarkvara installatsiooni kuupäev kujul "YYYY-MM-DD" või date objekt.

    Attributes:
        name (str): Tarkvara nimi.
        producer (str): Tarkvara tootja.
        install_date (datetime.date): Tarkvara installatsiooni kuupäev.
    """

    def __init__(self, name, producer, install_date):
        self.name = name
        self.producer = producer
        self.install_date = install_date  

    @property
    def install_date(self):
        """Tagastab tarkvara installatsiooni kuupäeva."""
        return self._install_date

    @install_date.setter
    def install_date(self, value):
        """
        Määrab installatsiooni kuupäeva, toetades nii stringi kui ka datetime.date tüüpi.
        Kui sisend on string, teisendab selle kuupäevaks.

        Args:
            value (str või datetime.date): Installatsiooni kuupäev.

        Raises:
            ValueError: Kui kuupäeva formaat ei ole "YYYY-MM-DD".
        """
        if isinstance(value, str):
            try:
                value = datetime.strptime(value, "%Y-%m-%d").date()
            except ValueError:
                raise ValueError("Installatsiooni kuupäev peab olema kujul YYYY-MM-DD")
        self._install_date = value

    def __str__(self):
        """Tagastab tarkvara info tekstina."""
        return f"{self.__class__.__name__}: {self.name}, {self.producer}, installitud: {self.install_date}"

class Freeware(Software):
    """
    Tasuta tarkvara klass, mis ei nõua täiendavaid atribuute ega meetodeid.
    """

    pass

class Shareware(Software):
    """
    Jagamistarkvara klass, mis sisaldab aegumiskuupäeva.

    Args:
        name (str): Tarkvara nimi.
        producer (str): Tarkvara tootja.
        install_date (str või datetime.date): Installatsiooni kuupäev.
        expiration_date (str või datetime.date): Aegumiskuupäev.

    Attributes:
        expiration_date (datetime.date): Tarkvara aegumiskuupäev.
    """

    def __init__(self, name, producer, install_date, expiration_date):
        super().__init__(name, producer, install_date)
        self.expiration_date = expiration_date

    @property
    def expiration_date(self):
        """Tagastab tarkvara aegumiskuupäeva."""
        return self._expiration_date

    @expiration_date.setter
    def expiration_date(self, value):
        """
        Määrab tarkvara aegumiskuupäeva, toetades nii stringi kui ka datetime.date tüüpi.

        Args:
            value (str või datetime.date): Aegumiskuupäev.

        Raises:
            ValueError: Kui kuupäeva formaat ei ole "YYYY-MM-DD".
        """
        if isinstance(value, str):
            try:
                value = datetime.strptime(value, "%Y-%m-%d").date()
            except ValueError:
                raise ValueError("Aegumiskuupäev peab olema kujul YYYY-MM-DD")
        self._expiration_date = value

    def __str__(self):
        """Tagastab tarkvara info koos aegumiskuupäevaga."""
        return super().__str__() + f", aegub: {self.expiration_date}"

class CommercialSoftware(Software):
    """
    Kommertstarkvara klass, mis lisaks aegumiskuupäevale sisaldab hinda.

    Args:
        name (str): Tarkvara nimi.
        producer (str): Tarkvara tootja.
        price (float): Tarkvara hind eurodes, peab olema mitte-negatiivne.
        install_date (str või datetime.date): Installatsiooni kuupäev.
        expiration_date (str või datetime.date): Aegumiskuupäev.

    Attributes:
        price (float): Tarkvara hind.
        expiration_date (datetime.date): Tarkvara aegumiskuupäev.
    """

    def __init__(self, name, producer, price, install_date, expiration_date):
        super().__init__(name, producer, install_date)
        self.price = price
        self.expiration_date = expiration_date

    @property
    def price(self):
        """Tagastab tarkvara hinna."""
        return self._price

    @price.setter
    def price(self, value):
        """
        Määrab tarkvara hinna.

        Args:
            value (float): Tarkvara hind.

        Raises:
            ValueError: Kui hind on negatiivne.
        """
        if value < 0:
            raise ValueError("Hind ei tohi olla negatiivne")
        self._price = value

    @property
    def expiration_date(self):
        """Tagastab tarkvara aegumiskuupäeva."""
        return self._expiration_date

    @expiration_date.setter
    def expiration_date(self, value):
        """
        Määrab tarkvara aegumiskuupäeva, toetades nii stringi kui ka datetime.date tüüpi.

        Args:
            value (str või datetime.date): Aegumiskuupäev.

        Raises:
            ValueError: Kui kuupäeva formaat ei ole "YYYY-MM-DD".
        """
        if isinstance(value, str):
            try:
                value = datetime.strptime(value, "%Y-%m-%d").date()
            except ValueError:
                raise ValueError("Aegumiskuupäev peab olema kujul YYYY-MM-DD")
        self._expiration_date = value

    def __str__(self):
        """Tagastab tarkvara info koos hinnaga ja aegumiskuupäevaga."""
        return super().__str__() + f", hind: {self.price}€, aegub: {self.expiration_date}"

class SoftwareManager:
    """
    Tarkvara haldusklass, mis võimaldab tarkvarasid lisada ja otsida.

    Attributes:
        software_list (List[Software]): Tarkvarade nimekiri.
    """

    def __init__(self):
        self.software_list: List[Software] = []

    def add(self, software: Software):
        """
        Lisab tarkvara haldusnimekirja.

        Args:
            software (Software): Tarkvara objekt.
        """
        self.software_list.append(software)

    def show_all(self):
        """Prindib välja kõik haldusnimekirjas olevad tarkvarad."""
        for sw in self.software_list:
            print(sw)

    def search_by_type(self, type_name):
        """
        Otsib tarkvarasid klassi nime põhjal (nt "Freeware", "Shareware", "CommercialSoftware").

        Args:
            type_name (str): Otsitava tarkvara tüüp.
        """
        for sw in self.software_list:
            if sw.__class__.__name__.lower() == type_name.lower():
                print(sw)

    def search_by_name_part(self, name_part):
        """
        Otsib tarkvarasid nime osalise vaste alusel.

        Args:
            name_part (str): Nime osa, mida otsida.
        """
        for sw in self.software_list:
            if name_part.lower() in sw.name.lower():
                print(sw)

    def search_by_producer(self, producer):
        """
        Otsib tarkvarasid tootja nime põhjal.

        Args:
            producer (str): Tootja nimi.
        """
        for sw in self.software_list:
            if producer.lower() == sw.producer.lower():
                print(sw)

    def usable_today(self):
        """
        Prindib välja kõik tarkvarad, mis on tänasel kuupäeval kasutatavad.

        Shareware ja CommercialSoftware puhul kontrollitakse aegumiskuupäeva,
        freeware-d prinditakse alati.
        """
        today = datetime.today().date()
        for sw in self.software_list:
            if isinstance(sw, (Shareware, CommercialSoftware)):
                if sw.expiration_date >= today:
                    print(sw)
            else:
                print(sw)

if __name__ == "__main__":
    manager = SoftwareManager()

    manager.add(Freeware("GIMP", "GNU", "2024-02-01"))
    manager.add(Shareware("WinZip", "Corel", "2024-01-15", "2025-01-01"))
    manager.add(CommercialSoftware("Adobe Photoshop", "Adobe", 120.0, "2023-11-01", "2025-12-31"))

    print("Kõik tarkvarad:")
    manager.show_all()

    print("\nTüübi järgi otsing: Shareware")
    manager.search_by_type("Shareware")

    print("\nNime osa järgi otsing: 'zip'")
    manager.search_by_name_part("zip")

    print("\nTootja järgi otsing: 'Adobe'")
    manager.search_by_producer("Adobe")

    print("\nKasutatavad täna:")
    manager.usable_today()
```

```python
from datetime import datetime, timedelta
import random
import pandas as pd
import string
from abc import ABC, abstractmethod

class Vehicle(ABC):
    """Abstraktne baas klass sõidukile parkimissüsteemis.

    Atribuudid:
        auto_number (str): Unikaalne sõiduki identifikaator.
        time_in (datetime): Sisenemise aeg parkimiskohta.
        time_out (datetime): Väljumise aeg parkimiskohast.
        fee (float): Parkimistasu, mis põhineb kestusel ja sõiduki tüübil.
    """
    def __init__(self, auto_number):
        self.auto_number = auto_number
        self.time_in = datetime.now()
        self.time_out = None
        self.__duration_minutes = None
        self.fee = None

    @abstractmethod
    def calculate_fee(self):
        """Arvuta parkimistasu. Tuleb realiseerida alamklassides."""
        pass

    def exit(self):
        """Määrab sõiduki väljumise aja."""
        self.time_out = datetime.now() + timedelta(minutes=random.randint(10, 100))

    @property
    def duration_minutes(self):
        """Arvutab parkimise aja minutites."""
        if self.time_out is None:
            self.exit()
        self.__duration_minutes = (self.time_out - self.time_in).total_seconds() / 60
        return self.__duration_minutes

    def __str__(self):
        return f'{self.__class__.__name__}, {self.auto_number}, {self.time_in.strftime("%d.%m.%Y %H:%M")}, {self.time_out.strftime("%d.%m.%Y %H:%M")}'

    @staticmethod
    def generate_car_number():
        """Genereerib juhusliku auto numbri, mis koosneb 3 numbrist ja 3 suurtähest."""
        digits = ''.join(random.choices(string.digits, k=3))
        letters = ''.join(random.choices(string.ascii_uppercase, k=3))
        return digits + letters

class NormalCar(Vehicle):
    """Klass tavalistele autodele koos tasu arvutamisega."""
    def calculate_fee(self):
        self.fee = round(2 * self.duration_minutes / 60, 2)
        return self.fee

class ElectricCar(Vehicle):
    """Klass elektriautodele koos tasu arvutamisega."""
    def calculate_fee(self):
        self.fee = round(3 * self.duration_minutes / 60 + 10, 2)
        return self.fee

class VIPCar(Vehicle):
    """Klass VIP-autodele koos  tasu arvutamisega."""
    def calculate_fee(self):
        self.fee = round(5 * self.duration_minutes / 60, 2)
        return self.fee

class ParkingSpot:
    """Esindab parkimiskohta, kuhu saab määrata kindlat tüüpi sõidukeid.

    Atribuudid:
        spot_ID (int): Parkimiskoha unikaalne identifikaator.
        spot_type (Vehicle): Sõiduki tüüp, mida koht lubab.
        occupied (bool): Näitab, kas koht on hõivatud.
    """
    def __init__(self, spot_ID: int, spot_type):
        self.spot_ID = spot_ID
        self.spot_type = spot_type
        self.occupied = False
        self.vehicle = None

    def assign_vehicle(self, vehicle):
        """Määrab sõiduki parkimiskohale, kui see on saadaval ja tüüp vastab."""
        if not self.occupied and isinstance(vehicle, self.spot_type):
            vehicle.time_in = datetime.now()
            self.vehicle = vehicle
            self.occupied = True
            print(f'Värav avatud. Sõiduk {vehicle.auto_number} {self.spot_type.__name__}')
            return True
        return False

    def remove_vehicle(self):
        """Eemaldab sõiduki parkimiskohast ja uuendab väljumise aja."""
        if self.occupied and self.vehicle:
            self.vehicle.exit()
            print(f'Sõiduk {self.vehicle.auto_number} {self.spot_type.__name__} eemaldatud kohalt {self.spot_ID}. Värav avatud.')
            self.occupied = False
            vehicle = self.vehicle
            self.vehicle = None
            return vehicle
        return None

    def __str__(self):
        status = 'Hõivatud' if self.occupied else 'Vaba'
        return f'Koht {self.spot_ID} {self.spot_type.__name__} - {status}'

car_numbers = set()
while len(car_numbers) < 10:
    car_numbers.add(Vehicle.generate_car_number())

car_types = [NormalCar, ElectricCar, VIPCar]
cars = [random.choice(car_types)(number) for number in car_numbers]


parking_spots = []
for i in range(1, 11):
    parking_spots.append(ParkingSpot(i, NormalCar))
for i in range(11, 16):
    parking_spots.append(ParkingSpot(i, ElectricCar))
for i in range(16, 18):
    parking_spots.append(ParkingSpot(i, VIPCar))


unparked_cars = []
for car in cars:
    parked = False
    for spot in parking_spots:
        if spot.assign_vehicle(car):
            parked = True
            break
    if not parked:
        unparked_cars.append(car)
        print(f'No available spots for {car.auto_number} {car.__class__.__name__}')


car_data = []
for spot in parking_spots:
    if spot.occupied:
        car = spot.remove_vehicle()
        car.calculate_fee()
        print(car)
        car_data.append([
            car.__class__.__name__,
            car.auto_number,
            car.time_in.strftime('%d.%m.%Y %H:%M'),
            car.time_out.strftime('%d.%m.%Y %H:%M'),
            round(car.duration_minutes),
            car.fee
        ])


"""Salvestab andmed CSV faili nimega 'parking_log.csv'.
Andmed sisaldavad sõiduki tüüpi, numbrit, sisenemise ja väljumise aega, parkimise kestust ja tasu."""

df = pd.DataFrame(car_data, columns = ['Type', 'Car Number', 'Time in', 'Time out', 'Duration (min)', 'Fee (eur)'])
df.to_csv('parking_log.csv', index = False)
df
```

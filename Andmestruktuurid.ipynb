Ülesanne 1 - Paroolide genereerimine ja turvalisus

Kirjeldus - Defineerisin paroolide genereerimise, ning baastaseme ja edasijõudnute algoritmi punktide määramise, et need töötaksid funktsioonidega.

```python
import random
import math

def parooli_märgid(arv):
    sümbolid = ("ABCDEFGHIJKLMNOPQRSTUVWXYZ"
                 "abcdefghijklmnopqrstuvwxyz"
                 "0123456789"
                 "!@#$%^&*()")
    return ''.join(random.choices(sümbolid, k=arv)) #''.join - muudab genereeritud paroolis sümbolid stringiks
```

```python
def parooli_tugevus(parool):
    punktid = 0
    
    if 8 <= len(parool) < 12:
        punktid += 2
    if len(parool) >= 12:
        punktid += 3
    if any(tähe_suurus.islower() for tähe_suurus in parool) and any(tähe_suurus.isupper() for tähe_suurus in parool):
        punktid += 2
    if any(number.isdigit() for number in parool):
        punktid += 2
    if any(sümbol in "!@#$%^&*()" for sümbol in parool):
        punktid += 3
    
    if punktid > 10:
        punktid = 10
    
    return punktid
```

```python
def entroopia(parool):
    sümbolite_arv = set(parool) # leiab erinevad märgid paroolis
    L = len(parool) # len annab erinevate märkide hulga
    N = len(sümbolite_arv)
    
    if N <= 1:
        return 1
    
    H = L * math.log2(N)
    if H < 10:
        return 4
    elif H < 20:
        return 6
    elif H < 30:
        return 8
    elif H < 40:
        return 9
    else:
        return 10
```

Paroolide loomine ja kriteeriumite kontrollimine, kust set välistab korduvad paroolid

```python
paroolid = set()
while len(paroolid) < 50:
    parooli_märkide_arv = random.randint(8, 15)
    parool = parooli_märgid(parooli_märkide_arv)
    
    if (any(tähe_suurus.isupper() for tähe_suurus in parool) and any(tähe_suurus.islower() for tähe_suurus in parool) and
        any(number.isdigit() for number in parool) and any(sümbol in "!@#$%^&*()" for sümbol in parool)):
        paroolid.add(parool)
```

Genereeritud paroolide hindamine ja väljastamine

```python
for parool in paroolid:
    Baastaseme_algoritm = parooli_tugevus(parool)
    Edasijõudnute_algoritm = entroopia(parool)
    
    print("Parool:", parool, " | Baastasel parooli punktid:", Baastaseme_algoritm, " | Entroopia turvalisuse skoor:", Edasijõudnute_algoritm)
    print()
    
    #Kasutasin "|"" visuaali jaoks :)
```

Kasutasin class funktsiooni, mis hoiab nii andmeid kui ka funktsionaalsust, kuna ma ei saanud def funktsiooniga hakkama. Leidsin class funktsiooni kohta internetist, kus kasutati just seda funktsiooni bussi süsteemi tegemiseks: https://stackoverflow.com/questions/17521203/creating-a-bus-system-in-python 

```python
class Peatus:
    def __init__(self, nimi, ooteaeg):
        self.nimi = nimi
        self.ooteaeg = ooteaeg

    def to_dict(self): # to_dict teeb sõnastikuks
        return {"Nimi": self.nimi, "ooteaeg": self.ooteaeg}
```

```python
class marsruut:
    def __init__(self, nimi, peatused, kava, koormus, ummikud):        #__init__ määrab alltoodu marsruudule
        self.nimi = nimi
        self.peatused = peatused
        self.kava = kava
        self.koormus = koormus
        self.ummikud =ummikud
```

```python
def to_dict(self):
        return {
            "nimi": self.nimi,
            "peatused": [peatus.to_dict() for peatus in self.peatused],
            "kava": self.kava,
            "koormus": self.koormus,
            "ummikud": self.ummikud
        }
```

```python
def arvuta_saabumisaeg(marsruut):
    saabumised = []
    aeg = 0
    for peatus in marsruut.peatused:
        liikumise_aeg = 3 * marsruut.ummikud
        aeg += liikumise_aeg + peatus.ooteaeg
        saabumised.append(round(aeg, 1))
    return saabumised
```

Peatused ja prognoositav saabumisaeg

```python
peatused_buss_5 = [
    Peatus("Kesklinn", 3),
    Peatus("Viru", 2),
    Peatus("Vabaduse väljak", 3),
    Peatus("Vineeri", 5),
    Peatus("Kalev", 4)
]

peatused_tramm_3 = [
    Peatus("Viru", 3),
    Peatus("Hobujaama", 4),
    Peatus("Tallinna ülikool", 6),
    Peatus("L.Koidula", 2),
    Peatus("Kadriorg", 4)
]

peatused_troll_2 = [
    Peatus("Viru", 3),
    Peatus("Mustamäe", 5),
    Peatus("Õismäe", 4),
    Peatus("Lasnamäe", 3),
    Peatus("Kesklinn", 2)
]
```

Algselt tegin programmi selliseks, mis väljastas nii tööpäevade kui ka nädalavahetuste ooteajad(ideega et kui on tööpäev siis praegune ooteaeg on selline: .... ja kui nädalavahetus siis praegune ooteaeg oleks teistsugune.), aga panin lõpuks ainult sõiduplaani.

```python
kava1 = {"Tööpäevad": "6:00 - 23:00, intervall 10 min", "Nädalavahetus": "8:00 - 22:00, intervall 15 min"}
kava2 = {"Tööpäevad": "5:00 - 24:00, intervall 12 min", "Nädalavahetus": "7:00 - 23:00, intervall 20 min"}
```

Marsruutude traspordid ja koormused

```python
marsruudud = [
    marsruut("Buss 5", peatused_buss_5, kava1, {"hommikul": 1.2, "pärval": 1.0, "õhtul": 1.3}, 1.0),
    marsruut("Tramm 3", peatused_tramm_3, kava2, {"hommikul": 1.1, "päeval": 1.3, "õhtul": 1.5}, 1.2),
    marsruut("Troll 2", peatused_troll_2, kava1, {"hommikul": 1.0, "päeval": 1.2, "õhtul": 1.3}, 1.1),
]
```

```python
for marsruut in marsruudud:
    marsruut.saabumised = arvuta_saabumisaeg(marsruut)
```

Arvutab mitu korda iga marsruut peatust läbib

```python
läbitud_peatused = {}
for marsruut in marsruudud:
    for peatus in marsruut.peatused:
        if peatus.nimi not in läbitud_peatused:
            läbitud_peatused[peatus.nimi] = 0
        läbitud_peatused[peatus.nimi] += 1
```

```python
populaarne_peatus = max(läbitud_peatused, key=läbitud_peatused.get)
maksimum = läbitud_peatused[populaarne_peatus]
```

```python
for marsruut in marsruudud:
    print("Marsruut:", marsruut.nimi)
    for i, peatus in enumerate(marsruut.peatused): # enumerate seob vastavad saabumisajad peatustega selleasemel et teeksin ise loendi (https://www.geeksforgeeks.org/enumerate-in-python/)
        print("Peatus:", peatus.nimi, ", saabumisaeg -", marsruut.saabumised[i], "min")
    print()

print("Kõige populaarsem peatus:", populaarne_peatus)
print("Kokku marsruute selles peatuses:", maksimum)
print()
print("Sõiduplaan:")
print("Buss 5 ja Tramm 2 kava:", kava1)
print("Troll 3 kava:", kava2)
```

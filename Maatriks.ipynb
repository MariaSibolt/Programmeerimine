1. Pythagorase ruut

```python
# Number stringiks

def get_digits(num):
    return [int(d) for d in str(num)]
```

```python
# Tühi sõnastik, võtmed 1-9
def maatriks(numbrid):
    pythagorase_ruut = {i: [] for i in range(1, 10)}
    for num in numbrid:
        if num in pythagorase_ruut:
            pythagorase_ruut[num].append(str(num))
    return {h: ' '.join(i) if i else '-' for h, i in pythagorase_ruut.items()}
```

```python
# Omaduste määramine
def omadused_määr(summa):
    if summa == 0:
        return "Omadus puudub või väga nõrk"
    elif summa == 1:
        return "Omadus on nõrgalt arenenud"
    elif summa == 2:
        return "Omadus on arenenud, kuid erinevates elusituatsioonides ebastabiilne"
    elif summa == 3:
        return "Omadus on arenenud normis"
    elif summa == 4:
        return "Omadus on kõrgelt arenenud"
    elif summa == 5:
        return "Omadus on maksimaalselt arenenud"
    else:
        return "Omadus on ülekoormatud"
```

```python
sünnikuupäev = input("Sisesta sünnikuupäev (DD.MM.YYYY): ")

# Kuu, päeva ja aasta eraldamine (.split('.'))
päev, kuu, aasta = sünnikuupäev.split('.')
a = päev + kuu + aasta

# Summade arvutamine
sum1 = sum(int(digit) for digit in a)
sum2 = sum(int(digit) for digit in str(sum1))
sum3 = sum1 - (int(päev[0]) * 2)
sum4 = sum(int(digit) for digit in str(sum3))
```

```python
# Pythagorase ruut
arvud = get_digits(a) + get_digits(sum1) + get_digits(sum2) + get_digits(sum3) + get_digits(sum4)
pythagorase_ruut = maatriks(arvud)
```

```python
# Pythagorase ruut tabelina joondatud / uurisin netist - https://docs.vultr.com/python/standard-library/str/center / Küsisin ka abi Chatgptlt
def psyhhomaatriks(pythagorase_ruut):
    max_len = max(len(i) for i in pythagorase_ruut.values())
    beautiful_print = {h: i.center(max_len) for h, i in pythagorase_ruut.items()}
    
    print(f"[{beautiful_print[1]} {beautiful_print[4]} {beautiful_print[7]}]")   # ma ei saanud automaatselt formattida ning panin järjest print
    print(f"[{beautiful_print[2]} {beautiful_print[5]} {beautiful_print[8]}]")
    print(f"[{beautiful_print[3]} {beautiful_print[6]} {beautiful_print[9]}]")
```

```python
# Number määratud omadusega
def omadused(pythagorase_ruut):
    c = {
        1: "Iseloomu tugevus - tahtejõud, sihikindlus",
        2: "Bioenergia - energia ja elujõud",
        3: "Huvid - teaduse ja tehnika vastu",
        4: "Tervis - jõud ja vastupidavus",
        5: "Loogiline mõtlemine, intuitsioon",
        6: "Töövõime - armastus füüsilise töö vastu",
        7: "Edukus ja vedamine",
        8: "Kohusetunne - vastutustunne ja korralikkus",
        9: "Vaimsete võimete tase, mälu, selgeltnägemise võime"
    }
    
    # Vastavalt numbri sagedusele omaduse tugevus
    print("\nDešifreerimine:")
    for key, desc in c.items():
        print(f"{desc}: {omadused_määr(len(pythagorase_ruut[key].split()))}")
```

```python
print("Pythagorase ruut:")
psyhhomaatriks(pythagorase_ruut)
omadused(pythagorase_ruut)
```

2. Vahemaade maatriks Eesti linnade vahel

```python
linnad = ["Tallinn", "Tartu", "Narva", "Pärnu", "Viljandi", "Rakvere"]

maatriks = [
    [0, 186, 210, 128, 159, 98],
    [186, 0, 182, 174, 73, 116],
    [210, 182, 0, 284, 235, 121],
    [128, 174, 284, 0, 77, 198],
    [159, 73, 235, 77, 0, 169],
    [98, 116, 121, 198, 169, 0]
]
```

```python
# Kaks linna, mille vahemaa on kõige väiksem
min_vahemaa = float('inf')
lähim_linn = []
for i in range(len(linnad)):
    for j in range(i + 1, len(linnad)):
        if 0 < maatriks[i][j] < min_vahemaa:
            min_vahemaa = maatriks[i][j]
            lähim_linn = [(linnad[i], linnad[j])]
        elif maatriks[i][j] == min_vahemaa:
            lähim_linn.append((linnad[i], linnad[j]))
```

```python
# Kaks linna, mille vahemaa on kõige suurem
max_vahemaa = 0
kaugeim_linn = []
for i in range(len(linnad)):
    for j in range(i + 1, len(linnad)):
        if maatriks[i][j] > max_vahemaa:
            max_vahemaa = maatriks[i][j]
            kaugeim_linn = [(linnad[i], linnad[j])]
        elif maatriks[i][j] == max_vahemaa:
            kaugeim_linn.append((linnad[i], linnad[j]))
```

```python
# Tsentraalseim linn 
summaarne_vahemaa = [sum(row) for row in maatriks]
tsentraalne = linnad[summaarne_vahemaa.index(min(summaarne_vahemaa))]
```

```python
# Kõige lähemal teistele 
min_max_vahemaa= [max(row) for row in maatriks]
kõige_lähim = linnad[min_max_vahemaa.index(min(min_max_vahemaa))]
```

```python
print("Maatriks:")
for row in maatriks:
    print(" ".join(f"{num:4}" for num in row))
print(" ")
print(f"Kõige lähemad linnad: {lähim_linn}, vahemaa: {min_vahemaa} km")
print(f"Kõige kaugemad linnad: {kaugeim_linn}, vahemaa: {max_vahemaa} km")
print(f"Kõige tsentraalsem linn: {tsentraalne}")
print(f"Kõige lähemal teistele olev linn: {kõige_lähim}")
```
